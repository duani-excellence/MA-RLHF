# CVALUES 

```python
# download from 'https://www.modelscope.cn/datasets/iic/CValues-Comparison/files'

dataset_path = './test.jsonl'
test_dataset = load_dataset('json', data_files=dataset_path)
print(test_dataset)

dataset_path = './train.jsonl'
train_dataset = load_dataset('json', data_files=dataset_path)
print(train_dataset)

train_dataset['test'] = test_dataset['train']
dataset = train_dataset
print(train_dataset)
```

```python
train_dataset.push_to_hub("xiaodongguaAIGC/CValues")
```

```python
def process_cvalues(examples):
    examples['chosen'] = examples['pos_resp']
    examples['rejected'] = examples['neg_resp']
    return examples
dataset = dataset.map(process_cvalues,
                            num_proc=8,
                            remove_columns = ["pos_resp",'neg_resp','pos_type','neg_type']
                           )
print(dataset)
```

```python

dataset.push_to_hub("xiaodongguaAIGC/CValues_DPO")
```
