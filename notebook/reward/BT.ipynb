```python
import numpy as np
from scipy.optimize import minimize

def bradley_terry(data, num_individuals):
    # 构建比较矩阵
    comparison_matrix = np.zeros((num_individuals, num_individuals))
    for i, j in data:
        comparison_matrix[i-1, j-1] = 1  # i胜j
        comparison_matrix[j-1, i-1] = -1  # j胜i

    # 定义似然函数
    def log_likelihood(x):
        logits = x.reshape((num_individuals, 1)) - x.reshape((1, num_individuals))
        probabilities = 1 / (1 + np.exp(-logits))
        log_probs = comparison_matrix * np.log(probabilities + 1e-10)
        return -np.sum(log_probs)

    # 初始估计值
    x0 = np.zeros(num_individuals)

    # 最大化似然函数
    result = minimize(log_likelihood, x0, method='BFGS')
    estimated_values = result.x

    return estimated_values

# 示例数据
# 假设有4个个体，数据表示1胜2、2胜3、3胜4、4胜1
data = [(1, 2), (2, 3), (3, 4), (1, 3)]
num_individuals = 4

# 计算能力或优势值
estimated_values = bradley_terry(data, num_individuals)

print("个体的能力或优势值:")
for i in range(num_individuals):
    print(f"个体 {i+1}: {estimated_values[i]}")
```
